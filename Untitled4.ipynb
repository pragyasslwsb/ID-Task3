```python

np.random.seed(42)

n = 500

pages = [
    "Home", "Products", "Pricing",
    "Blog", "Contact", "Checkout"
]

df = pd.DataFrame({
    "Session_ID": range(1, n + 1),
    "User_ID": np.random.randint(1001, 1201, n),
    "Traffic_Source": np.random.choice(
        ["Organic", "Direct", "Social", "Referral"],
        n
    ),
    "Landing_Page": np.random.choice(
        ["Home", "Products", "Blog", "Pricing"], n
    ),
    "Pages_Viewed": np.random.randint(1, 9, n),
    "Session_Duration_Min": np.round(
        np.random.uniform(0.2, 15, n), 2
    ),
    "Exit_Page": np.random.choice(pages, n),
    "Converted": np.random.choice(
        ["Yes", "No"], n, p=[0.25, 0.75]
    )
})

df.head()
```

```python

print("Dataset shape:", df.shape)

print("\nColumn information:")
df.info()

print("\nMissing values:")
print(df.isnull().sum())

print("\nDuplicate rows:", df.duplicated().sum())

df.describe()
```

```python

print("Dataset shape:", df.shape)

print("\nColumn information:")
df.info()

print("\nMissing values:")
print(df.isnull().sum())

print("\nDuplicate rows:", df.duplicated().sum())

df.describe()
```

```python

total_sessions = df["Session_ID"].nunique()
unique_users = df["User_ID"].nunique()

total_page_views = df["Pages_Viewed"].sum()

avg_session_duration = df[
    "Session_Duration_Min"
].mean()

avg_pages_per_session = df[
    "Pages_Viewed"
].mean()

conversion_rate = (
    (df["Converted"] == "Yes").mean() * 100
)

print("Total Sessions:", total_sessions)
print("Unique Users:", unique_users)
print("Estimated Page Views:", total_page_views)
print("Average Session Duration:",
      round(avg_session_duration, 2), "minutes")
print("Average Pages per Session:",
      round(avg_pages_per_session, 2))
print("Conversion Rate:", round(conversion_rate, 2), "%")
```

```python

sns.set_theme(style="whitegrid")

# 1. Traffic by source
plt.figure(figsize=(8, 5))
sns.countplot(data=df, x="Traffic_Source")
plt.title("Website Sessions by Traffic Source")
plt.xlabel("Traffic Source")
plt.ylabel("Number of Sessions")
plt.show()

# 2. Session duration distribution
plt.figure(figsize=(8, 5))
sns.histplot(data=df, x="Session_Duration_Min", bins=20)
plt.title("Session Duration Distribution")
plt.xlabel("Duration (minutes)")
plt.show()

# 3. Pages viewed vs conversion
plt.figure(figsize=(8, 5))
sns.barplot(
    data=df,
    x="Converted",
    y="Pages_Viewed",
    errorbar=None
)
plt.title("Average Pages Viewed by Conversion Status")
plt.show()

# 4. Exit pages
plt.figure(figsize=(9, 5))
sns.countplot(
    data=df,
    y="Exit_Page",
    order=df["Exit_Page"].value_counts().index
)
plt.title("Sessions Ending on Each Page")
plt.xlabel("Number of Sessions")
plt.show()
```

```python

exit_analysis = (
    df.groupby("Exit_Page")
    .agg(
        Sessions=("Session_ID", "count"),
        Avg_Duration=("Session_Duration_Min", "mean"),
        Avg_Pages_Viewed=("Pages_Viewed", "mean"),
        Conversions=("Converted",
                     lambda x: (x == "Yes").sum())
    )
    .sort_values("Sessions", ascending=False)
)

exit_analysis["Conversion_Rate_%"] = (
    exit_analysis["Conversions"]
    / exit_analysis["Sessions"] * 100
).round(2)

display(exit_analysis.round(2))
```
